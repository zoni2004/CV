```python
#TASK1 - Object-Oriented Grocery Management
class GroceryManager:
    def __init__(self):
        self.grocery_list = {}

    def add_item(self, item, quantity, price):
        if not isinstance(item, str) or not item.strip():
            print("Error: Item name must be a non-empty string.")
            return
        if not isinstance(quantity, (int, float)) or quantity <= 0:
            print("Error: Quantity must be a positive number.")
            return
        if not isinstance(price, (int, float)) or price <= 0:
            print("Error: Price must be a positive number.")
            return

        self.grocery_list[item.strip()] = {"quantity": quantity, "price": price}
        print(f"Added/Updated: {quantity} x {item.strip()} at ${price} each.")

    def remove_item(self, item):
        if not isinstance(item, str) or not item.strip():
            print("Error: Item name must be a non-empty string to remove.")
            return

        item_to_remove = item.strip()
        if item_to_remove in self.grocery_list:
            del self.grocery_list[item_to_remove]
            print(f"Removed '{item_to_remove}' from the list.")
        else:
            print(f"Error: '{item_to_remove}' not found in the grocery list.")

    def view_list(self):
        if not self.grocery_list:
            print("Grocery list is empty.")
            return

        print("\nCurrent Grocery List:")
        for item, details in self.grocery_list.items():
            print(f"- {item}: Quantity={details['quantity']}, Price=${details['price']:.2f}")

    def calculate_total(self):
        total_cost = 0
        for item, details in self.grocery_list.items():
            total_cost += details['quantity'] * details['price']
        print(f"\nTotal cost of all items: ${total_cost:.2f}")
        return total_cost


manager = GroceryManager()


manager.add_item("Apples", 5, 1.50)
manager.add_item("Milk", 1, 3.00)
manager.add_item("Bread", 2, 2.25)
manager.add_item("Cheese", 1, 5.00)

manager.view_list()
manager.calculate_total()
manager.remove_item("Eggs")
manager.remove_item("Bread")
manager.view_list()
manager.calculate_total()

#invalid inputs
manager.add_item("", 1, 1.00)
manager.add_item("Oranges", -2, 1.00)
manager.add_item("Grapes", 1, -0.50)
```

```python
#Task2 - Advanced Student Record System
def get_student_with_highest_average_grade(student_records):
    if not student_records:
        return "No students in the record system."

    highest_avg_grade = -1
    student_with_highest_avg = ""

    for student_id, data in student_records.items():
        grades = data.get('Grades', [])
        if grades:
            current_avg = sum(grades) / len(grades)
            if current_avg > highest_avg_grade:
                highest_avg_grade = current_avg
                student_with_highest_avg = data.get('Name', f"Unknown Student (ID: {student_id})")

    if not student_with_highest_avg:
        return "No students with grades found."

    return student_with_highest_avg

def search_students_by_major(student_records, major):
    found_students = []
    for student_id, data in student_records.items():
        if data.get('Major') == major:
            found_students.append(data.get('Name', f"Unknown Student (ID: {student_id})"))

    if found_students:
        print(f"Students enrolled in {major}:")
        for student_name in found_students:
            print(f"- {student_name}")
    else:
        print(f"No students found in {major} major.")


student_records = {
        'S001': {'Name': 'Alice Smith', 'Major': 'Computer Science', 'Grades': [95, 88, 92, 70]},
        'S002': {'Name': 'Bob Johnson', 'Major': 'Electrical Engineering', 'Grades': [80, 82, 79]},
        'S003': {'Name': 'Charlie Brown', 'Major': 'Computer Science', 'Grades': [75, 80, 85, 90]},
        'S004': {'Name': 'Diana Prince', 'Major': 'Physics', 'Grades': [98, 97, 99]},
        'S005': {'Name': 'Eve Adams', 'Major': 'Electrical Engineering', 'Grades': [85, 87, 83]}
    }

print("\n--- Student Record System Demo ---")

highest_avg_student = get_student_with_highest_average_grade(student_records)
print(f"Student with the highest average grade: {highest_avg_student}")
search_students_by_major(student_records, 'Computer Science')
search_students_by_major(student_records, 'Physics')
search_students_by_major(student_records, 'Mathematics')
```

```python
#Task4 - Interactive Geometry & Blank Canvas

import numpy as np
import cv2
import matplotlib.pyplot as plt

image_size = 800
canvas = np.zeros((image_size, image_size, 3), dtype=np.uint8)

center_x, center_y = image_size // 2, image_size // 2
print(f"Center coordinates of the canvas: ({center_x}, {center_y})")

color1 = (0, 0, 255)
color2 = (255, 0, 0)

radii = [50, 100, 150, 200, 250]

for i, r in enumerate(radii[::-1]):
    color = color1 if (i % 2 == 0) else color2
    cv2.circle(canvas, (center_x, center_y), r, color, -1)

outermost_radius = radii[-1]

x1 = center_x - outermost_radius
y1 = center_y - outermost_radius
x2 = center_x + outermost_radius
y2 = center_y + outermost_radius

bounding_box_color = (0, 255, 0)
cv2.rectangle(canvas, (x1, y1), (x2, y2), bounding_box_color, 2)

plt.figure(figsize=(10, 10))
plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
plt.title('Target Board with Concentric Circles and Bounding Box', fontsize=16, color='navy', pad=15)
plt.axis('off')
plt.show()
```

```python
#Task5 - Advanced Blurring & NumPy ROI Extraction

image_path = '/content/Screenshot 2026-08-18 200017.png'
img = cv2.imread(image_path)

if img is None:
    print(f"Error: Image not found at {image_path}.")
else:
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    blurred_img = cv2.GaussianBlur(img_rgb, (25, 25), 0)

    h, w, _ = img_rgb.shape
    center_x, center_y = w // 2, h // 2

    roi_size = 300
    half_roi = roi_size // 2

    x1 = center_x - half_roi
    y1 = center_y - half_roi
    x2 = center_x + half_roi
    y2 = center_y + half_roi

    roi_original = img_rgb[y1:y2, x1:x2]
    roi_blurred = blurred_img[y1:y2, x1:x2]

    plt.figure(figsize=(12, 6))

    plt.subplot(1, 2, 1)
    plt.imshow(roi_original)
    plt.title('Original Image ROI', fontsize=14, color='darkgreen')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(roi_blurred)
    plt.title('Blurred Image ROI (25x25 Gaussian)', fontsize=14, color='darkred')
    plt.axis('off')

    plt.tight_layout()
    plt.show()
```

```python
#Task6 - Alpha Blending & Typography

image_path = '/content/Screenshot 2026-08-18 200017.png'
original_img = cv2.imread(image_path)

if original_img is None:
    print(f"Error: Image not found at {image_path}.")
else:
    original_img_rgb = cv2.cvtColor(original_img, cv2.COLOR_BGR2RGB)

    h, w, _ = original_img.shape

    overlay_height = int(h * 0.20)
    overlay_start_y = h - overlay_height

    overlay = np.zeros_like(original_img, dtype=np.uint8)

    blue_color = (255, 0, 0)
    cv2.rectangle(overlay, (0, overlay_start_y), (w, h), blue_color, -1)

    # alpha: weight of the first image elements.
    # beta: weight of the second image elements.
    # gamma: scalar added to each sum.
    alpha = 1.0
    beta = 0.5
    blended_img = cv2.addWeighted(original_img, alpha, overlay, beta, 0)

    blended_img_rgb = cv2.cvtColor(blended_img, cv2.COLOR_BGR2RGB)

    text = "King of Curses"
    font = cv2.FONT_HERSHEY_SIMPLEX
    font_scale = 1
    font_thickness = 3
    text_color = (255, 255, 255)

    text_size = cv2.getTextSize(text, font, font_scale, font_thickness)[0]
    text_x = (w - text_size[0]) // 2
    text_y = overlay_start_y + (overlay_height + text_size[1]) // 2

    cv2.putText(blended_img_rgb, text, (text_x, text_y), font, font_scale, text_color, font_thickness, cv2.LINE_AA)

    plt.figure(figsize=(10, 10))
    plt.imshow(blended_img_rgb)
    plt.title('Image with Semi-Transparent Overlay and Text', fontsize=16, color='darkblue', pad=15)
    plt.axis('off')
    plt.show()
```

```python
#Task7 - Matrix-Based Rotation & Adaptive Thresholding

image_path_grayscale = '/content/grayscale-image.jpg'
img_gray = cv2.imread(image_path_grayscale, cv2.IMREAD_GRAYSCALE)

if img_gray is None:
    print(f"Error: Grayscale image not found at {image_path_grayscale}.")
else:
    ret, global_thresh = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY)

    adaptive_thresh = cv2.adaptiveThreshold(img_gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, \
                                            cv2.THRESH_BINARY, 11, 2)

    plt.figure(figsize=(15, 7))
    plt.subplot(1, 2, 1)
    plt.imshow(global_thresh, cmap='gray')
    plt.title('Global Binary Thresholding', fontsize=14, color='navy')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(adaptive_thresh, cmap='gray')
    plt.title('Adaptive Gaussian Thresholding', fontsize=14, color='darkgreen')
    plt.axis('off')
    plt.tight_layout()
    plt.show()

    (h, w) = adaptive_thresh.shape[:2]
    center = (w // 2, h // 2)
    angle = 45
    scale = 0.8

    rotation_matrix = cv2.getRotationMatrix2D(center, angle, scale)

    rotated_img = cv2.warpAffine(adaptive_thresh, rotation_matrix, (w, h))

    plt.figure(figsize=(8, 8))
    plt.imshow(rotated_img, cmap='gray')
    plt.title('Adaptive Threshold Output Rotated 45 Degrees (Scaled 0.8)', fontsize=14, color='purple')
    plt.axis('off')
    plt.show()
```

```python
#Task8 - Masking with Bitwise Logi


image_path1 = '/content/Screenshot 2026-08-18 200017.png'
image_path2 = '/content/grayscale-image.jpg'
common_size = (500, 500)

img1 = cv2.imread(image_path1)
img2_gray = cv2.imread(image_path2, cv2.IMREAD_GRAYSCALE)

if img1 is None:
    print(f"Error: Image 1 not found at {image_path1}.")
elif img2_gray is None:
    print(f"Error: Image 2 (grayscale) not found at {image_path2}.")
else:
    img2 = cv2.cvtColor(img2_gray, cv2.COLOR_GRAY2BGR)

    img1_resized = cv2.resize(img1, common_size)
    img2_resized = cv2.resize(img2, common_size)

    mask = np.zeros(common_size, dtype=np.uint8)

    center = (common_size[0] // 2, common_size[1] // 2)
    radius = 150
    thickness = -1
    cv2.circle(mask, center, radius, 255, thickness)

    fg = cv2.bitwise_and(img1_resized, img1_resized, mask=mask)

    mask_inv = cv2.bitwise_not(mask)

    bg = cv2.bitwise_and(img2_resized, img2_resized, mask=mask_inv)

    final_image = cv2.bitwise_or(fg, bg)

    plt.figure(figsize=(15, 10))

    plt.subplot(2, 3, 1)
    plt.imshow(cv2.cvtColor(img1_resized, cv2.COLOR_BGR2RGB))
    plt.title('Image 1 (Resized)', fontsize=12)
    plt.axis('off')

    plt.subplot(2, 3, 2)
    plt.imshow(cv2.cvtColor(img2_resized, cv2.COLOR_BGR2RGB))
    plt.title('Image 2 (Resized)', fontsize=12)
    plt.axis('off')

    plt.subplot(2, 3, 3)
    plt.imshow(mask, cmap='gray')
    plt.title('Mask', fontsize=12)
    plt.axis('off')

    plt.subplot(2, 3, 4)
    plt.imshow(cv2.cvtColor(fg, cv2.COLOR_BGR2RGB))
    plt.title('Image 1 with Mask', fontsize=12)
    plt.axis('off')

    plt.subplot(2, 3, 5)
    plt.imshow(mask_inv, cmap='gray')
    plt.title('Inverted Mask', fontsize=12)
    plt.axis('off')

    plt.subplot(2, 3, 6)
    plt.imshow(cv2.cvtColor(final_image, cv2.COLOR_BGR2RGB))
    plt.title('Final Combined Image', fontsize=12)
    plt.axis('off')

    plt.tight_layout()
    plt.show()
```
